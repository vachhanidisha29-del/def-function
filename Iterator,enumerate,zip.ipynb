```python
numbers = [10, 20, 30, 40, 50]

iterator = iter(numbers)

for number in iterator:
    print(number)
```

```python
numbers = [10, 20, 30, 40, 50]

iterator = iter(numbers)

while True:
    try:
        print(next(iterator))
    except StopIteration:
        break
```

```python
numbers = [10, 20, 30, 40, 50]

iterator = iter(numbers)

total = 0

while True:
    try:
        total += next(iterator)
    except StopIteration:
        break

print("Sum:", total)
```

```python
names = ["Rahul", "Priya", "Amit", "Neha"]

for index, name in enumerate(names):
    print(index, name)
```

```python
names = ["Rahul", "Priya", "Amit", "Neha"]

for index, name in enumerate(names, start=1):
    print(index, name)
```

```python
languages = ["Java", "C++", "Python", "JavaScript", "SQL"]

for index, language in enumerate(languages, start=1):
    if language == "Python":
        print("Python is at position:", index)
```

```python
names = ["Rahul", "Priya", "Amit"]
marks = [80, 90, 75]

result = list(zip(names, marks))

print(result)
```

```python

names = ["Rahul", "Priya", "Amit"]
marks = [80, 90, 75]

student_marks = dict(zip(names, marks))

print(student_marks)
```

```python
names = ["Rahul", "Priya", "Amit", "Neha"]
marks = [80, 90, 75, 85]

for name, mark in zip(names, marks):
    print(name, ":", mark)
```
