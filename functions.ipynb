```python
def square(num):
    return num * num

print(square(5))
```

```python
def maximum(a, b):
    if a > b:
        return a
    else:
        return b

print(maximum(10, 20))
```

```python
def greet(name="Student"):
    print("Hello", name)

greet("Disha,shiv")
greet()
```

```python
def largest(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

print(largest(10, 25, 15))
```

```python
def check_number(num):
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"

print(check_number(-5))
```

```python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32

print(celsius_to_fahrenheit(25))
```

```python
def count_vowels(text):
    count = 0

    for char in text:
        if char.lower() in "aeiou":
            count += 1

    return count

print(count_vowels("Python Programming"))
```

```python
def reverse_string(text):
    return text[::-1]

print(reverse_string("Python"))
```

```python
def check_palindrome(text):
    if text == text[::-1]:
        return "Palindrome"
    else:
        return "Not Palindrome"

print(check_palindrome("madam"))
print(check_palindrome("python"))
```
