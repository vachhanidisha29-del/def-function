```python
def average_marks(marks):
    return sum(marks) / len(marks)


marks = [80, 70, 90, 60, 75]
print(average_marks(marks))
```

```python
def find_max(a, b):
    if a > b:
        return a
    else:
        return b


print(find_max(25, 40))
```

```python
def calculate_total_sales(sales):
    return sum(sales)


sales = [10000, 15000, 12000, 18000]
print(calculate_total_sales(sales))
```

```python
def count_even(numbers):
    count = 0

    for number in numbers:
        if number % 2 == 0:
            count += 1

    return count


numbers = [1, 2, 3, 4, 5, 6, 8, 10]
print(count_even(numbers))
```

```python
def calculate_discount(price, rate=10):
    discount = price * rate / 100
    final_price = price - discount

    return final_price


print(calculate_discount(1000))
```

```python
def remove_duplicates(data):
    new_list = []

    for item in data:
        if item not in new_list:
            new_list.append(item)

    return new_list


data = [10, 20, 10, 30, 20, 40, 30]
print(remove_duplicates(data))
```

```python
def sales_summary(sales):
    total = sum(sales)
    average = total / len(sales)
    highest = max(sales)

    return total, average, highest


sales = [10000, 15000, 12000, 18000, 20000]

total, average, highest = sales_summary(sales)

print("Total Sales:", total)
print("Average Sales:", average)
print("Highest Sale:", highest)
```

```python
def calculate_bonus(salary):
    if salary > 50000:
        return salary * 10 / 100
    else:
        return salary * 5 / 100


print(calculate_bonus(60000))
print(calculate_bonus(40000))
```
