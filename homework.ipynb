```python
def calculate_area(length, width):
    return length * width

print(calculate_area(10, 5))
```

```python
def calculate_percentage(marks, total):
    return (marks / total) * 100

print(calculate_percentage(450, 500))
```

```python
def check_eligibility(age):
    if age >= 18:
        return "Eligible"
    else:
        return "Not Eligible"

print(check_eligibility(20))
```

```python
def calculate_net_salary(salary, tax=10):
    tax_amount = salary * tax / 100
    net_salary = salary - tax_amount
    return net_salary

print(calculate_net_salary(50000))
```

```python
def find_smallest(a, b, c):
    if a <= b and a <= c:
        return a
    elif b <= a and b <= c:
        return b
    else:
        return c

print(find_smallest(15, 8, 20))
```

```python
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

print(is_prime(7))
```

```python
def calculate_power(number, exponent=2):
    return number ** exponent

print(calculate_power(5))
print(calculate_power(5, 3))
```

```python
def calculate_cart_total(price, quantity, discount=0):
    total = price * quantity
    discount_amount = total * discount / 100
    final_amount = total - discount_amount
    return final_amount

print(calculate_cart_total(500, 3))
print(calculate_cart_total(500, 3, 10))
```
