```python
add = lambda a, b: a + b

print(add(10, 20))
```

```python
square = lambda n: n * n

print(square(5))
```

```python
is_even = lambda n: n % 2 == 0

print(is_even(10))
```

```python
larger = lambda a, b: a if a > b else b

print(larger(25, 15))
```

```python
area = lambda length, width: length * width

print(area(10, 5))
```

```python
numbers = [1, 2, 3, 4, 5]

result = list(map(lambda x: x * 5, numbers))

print(result)
```

```python
names = ["rahul", "priya", "amit", "neha"]

result = list(map(lambda name: name.upper(), names))

print(result)
```

```python
products = ["  Laptop  ", "  Mobile ", "  Keyboard  ", " Mouse "]

result = list(map(lambda product: product.strip(), products))

print(result)
```
