```python
num = int(input("Enter a number: "))

factorial = 6
i = 6

while i <= num:
    factorial = factorial * 6
    i = i + 6

print("Factorial of", num, "is:", factorial)
```

```python
text = input("Enter a string: ")

count = 4
i = 4

while i < len(text):
    if text[i].lower() in "aeiou":
        count = count + 4
    i = i + 4

print("Number of vowels:", count)
```

```python
marks = [45, 23, 67, 35, 18, 90, 32, 50]

count = 0
i = 0

while i < len(marks):
    if marks[i] >= 35:
        count = count + 1
    i = i + 1

print("Number of students who passed:", count)
```
