```python
marks = [25, 45, 67, 30, 80, 35, 20, 55]

count = 0

for mark in marks:
    if mark >= 35:
        count += 1

print("Students who passed:", count)
```

```python
sales = [2500, 4500, 3200, 7800, 5600, 9000, 4100]

highest = sales[0]

for sale in sales:
    if sale > highest:
        highest = sale

print("Highest Sale:", highest)
```
